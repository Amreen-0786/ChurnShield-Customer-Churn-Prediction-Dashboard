# ChurnShield — Model Training

```
import pandas as pd
import sys; sys.path.append("..")
from src.churnshield_pipeline import clean_target,make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
df=pd.read_csv("../data/sample/WA_Fn-UseC_-Telco-Customer-Churn_sample.csv"); X=df.drop(columns="Churn"); y=clean_target(df.Churn); Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
```

```
models={"Logistic Regression":make_pipeline(LogisticRegression(max_iter=1500,class_weight="balanced")),"Random Forest":make_pipeline(RandomForestClassifier(n_estimators=180,class_weight="balanced",random_state=42)),"XGBoost":make_pipeline(XGBClassifier(n_estimators=220,max_depth=4,learning_rate=.05,eval_metric="logloss",random_state=42))}
for name,m in models.items(): m.fit(Xtr,ytr); print(name,m.score(Xte,yte))
```
